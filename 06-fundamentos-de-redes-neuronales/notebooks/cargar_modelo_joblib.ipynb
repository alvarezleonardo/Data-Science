# Cargar un modelo guardado con `joblib`

**Módulo 06 — Fundamentos de redes neuronales.** Recurso de las clases 23-24, sobre persistencia de modelos.

**Requisito**: este notebook necesita el archivo `mi_modelo.joblib`, que **no está versionado** en el repositorio (los artefactos binarios están en `.gitignore`). Hay que ejecutar primero `guardar_modelo_joblib_pickle.ipynb` para generarlo antes de correr esta notebook.

```python
import joblib
import numpy as np
```

## Cargar el modelo

`joblib.load(archivo)` deserializa el archivo y devuelve el objeto original: el `MLPClassifier` ya entrenado, con todos sus pesos. A partir de acá, `mlp` se usa exactamente igual que un modelo recién entrenado con `.fit()` — no hace falta ni posible volver a entrenarlo, ya viene listo para predecir.

**Nota sobre compatibilidad de versiones**: un modelo guardado con una versión de scikit-learn puede fallar al cargarse, o comportarse distinto, con otra versión de la librería. Conviene registrar junto al modelo la versión de scikit-learn (y de sus dependencias) con la que fue entrenado.

**Advertencia de seguridad:** `joblib` usa `pickle` por debajo, así que hereda su riesgo — cargar un archivo de origen desconocido puede ejecutar código arbitrario. Vale lo mismo que para `.pkl`: solo cargar modelos de origen confiable.

```python
#Cargar el modelo entrenado
mlp = joblib.load('mi_modelo.joblib')
```

## Predecir con el modelo cargado

Cada predicción se hace sobre un único ejemplo, representado como un array de NumPy de dos valores: `sepal length` y `petal length`, en centímetros y en ese mismo orden — el orden con el que se entrenó el modelo. Respetar ese orden es obligatorio: si se invierten los atributos, el modelo predice sobre datos que no tienen el significado que espera.

`.reshape(1, -1)` es necesario porque scikit-learn espera una matriz 2D de forma `(cantidad de muestras, cantidad de atributos)`, no un vector suelto de una dimensión. El `1` indica una sola muestra, y el `-1` le dice a NumPy que infiera automáticamente la cantidad de columnas (en este caso, 2).

```python
#Predecir
print(mlp.predict(np.array([6.1, 4.7]).reshape(1, -1)))
print(mlp.predict(np.array([5.7, 1.7]).reshape(1, -1)))
print(mlp.predict(np.array([7.7, 6.9]).reshape(1, -1)))
```
