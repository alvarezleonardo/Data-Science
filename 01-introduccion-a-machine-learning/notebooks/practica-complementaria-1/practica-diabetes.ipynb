```python
import numpy as np
unique_elements, counts_elements_train = np.unique(y_train, return_counts=True)

percentages_train = (counts_elements_train / y_train.size) * 100
```

```python
print("Unique elements in y_train: ", unique_elements)
print("Counts of each unique element in y_train: ", counts_elements_train)
print("Percentages of each unique element in y_train: ", percentages_train)
print(y_train.shape)
```

```python
unique_elements_test, counts_elements_test = np.unique(y_test, return_counts=True)
percentages_test = (counts_elements_test / y_test.size) * 100
```

```python
print("Unique elements in y_test: ", unique_elements_test)
print("Counts of each unique element in y_test: ", counts_elements_test)
print("Percentages of each unique element in y_test: ", percentages_test)   
print(y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
print(predictions)
```

```python
print(y_test.values)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, predictions)
print("Confusion Matrix:\n", cm)
```

```python
plt.figure(figsize=(4,3))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix Heatmap')
plt.show()
```

```python
from sklearn.metrics import accuracy_score, classification_report 

print("Accuracy: ", accuracy_score(y_test, predictions))

print("Classification Report:\n", classification_report(y_test, predictions))
```

```python
print(classification_report(y_test, predictions,target_names=["No Diabetes", "Diabetes"]))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score 

print("Precision: ", precision_score(y_test, predictions, average=None))
print("Recall: ", recall_score(y_test, predictions, average=None))
print("F1 Score: ", f1_score(y_test, predictions, average=None))
```

```python
y_scores = model.predict_proba(X_test)
print("Predicted probabilities for the positive class:\n", y_scores)    

pd.DataFrame({'proba_0':y_scores[:,0], 'proba_1':y_scores[:,1],'categoria':predictions}).head(15)
```

```python
from sklearn.metrics import roc_curve
from sklearn.metrics import confusion_matrix

import matplotlib
import matplotlib.pyplot as plt
%matplotlib inline

fpr, tpr, thresholds = roc_curve(y_test, y_scores[:,1])
```

```python
from sklearn.metrics import roc_auc_score

auc = roc_auc_score(y_test, y_scores[:,1])
print("AUC Score: ", auc)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
import numpy as np
```

```python
numeric_features = [0,1,2,3,4,5,6]
numeric_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())])

categorical_features = [7]
categorical_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore'))])
```

```python
processor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)])   
```

```python
pipeline = Pipeline(steps=[('preprocessor', processor),
                           ('classifier', LogisticRegression())])

model = pipeline.fit(X_train, y_train)
print(model)
```

```python
predictions = model.predict(X_test)
y_scores = model.predict_proba(X_test)
print("Predicted probabilities for the positive class:\n", y_scores)

cm=confusion_matrix(y_test, predictions)
print("Confusion Matrix:\n", cm)


```
